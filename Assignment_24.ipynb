```python
from langchain_ollama import ChatOllama

chat_model = ChatOllama(
    model="llama3.2:1b",
    temperature=0.7
)

print("Ollama chat model initialized successfully!")
```

```python
response = chat_model.invoke("names of the 5 top run scorers in international cricket")

print(response.content)
```

### observation

- Successfully installed and configured Ollama with LangChain using the Llama 3.2 (1B) model.
- The model was invoked using ChatOllama and generated a response to a natural-language prompt.
- This confirmed successful local LLM integration.
- The generated response also demonstrated that smaller language models can produce factual inaccuracies.

```python
import os
import langsmith

from langsmith import Client

print("LangSmith version:", langsmith.__version__)
print("Tracing enabled:", os.getenv("LANGSMITH_TRACING"))
print("API key available:", bool(os.getenv("LANGSMITH_API_KEY")))
print("Project:", os.getenv("LANGSMITH_PROJECT"))

client = Client()

try:
    projects = list(client.list_projects())
    print("\nSuccessfully connected to LangSmith.")
    print("Number of accessible projects:", len(projects))
except Exception as e:
    print("\nLangSmith connection failed:")
    print(type(e).__name__, str(e))
```
